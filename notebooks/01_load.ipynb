```python
# imports 
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle
import os

# interactive plots 
%matplotlib widget

# if the working directory is notebooks/, move to the project root
if os.getcwd().endswith("notebooks"):
    os.chdir("..")

DATA_FILE   = "data/001MoDe_R1.csv"
MARKER_FILE = "data/001MoDe_R1.marker.csv"
```

```python
from src import io_mouse as mm
```

## Header (line 1 of the file, `key value;key value;...`)

```python
header = mm.read_header(DATA_FILE)
for k, v in header.items():
    print(f"  {k}: {v}")
assert header["externalRadius"] == 250.0 and header["taskRadius"] == 209.5
```

## Samples

Raw columns: `timestamp, mouseX, mouseY, mouseInTarget`, in screen pixels (y points down). `read_data` re-centers on the task center (`centerX`, `centerY`) and flips y, as in the expected figures.

```python
data = mm.read_data(DATA_FILE, header)
print(len(data["x"]), "samples")
print("first sample: x =", data["x"][0], " y =", data["y"][0])
```

## Markers and record windows

```python
markers, ref_table = mm.read_markers(MARKER_FILE)
records = mm.get_records(markers)
print("Number of records:", len(records))  # expected: 5
t0 = data["t_ms"][0]  # reference time = first sample
for a, b in records:
    print(f"{(a - t0)/1000:8.3f} to {(b - t0)/1000:8.3f}  ({(b - a)/1000:.3f} s)")
# expected: 0.004 to 20.007 (20.003 s), 40.011 to 60.022 (20.011 s), ... 160.090 to 180.104 (20.014 s)
```
